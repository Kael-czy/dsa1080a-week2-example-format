## 1. Load libraries and data

Standard imports and loading the dataset produced in Week 1.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# Illustrative: path to the dataset used in Week 1 (e.g. star_classification.csv)
DATA_PATH = "data/star_classification.csv"
df = pd.read_csv(DATA_PATH)
df.head()
```

## 2. Handle missing values and outliers

Check for nulls, then use an IQR-based rule to flag extreme outliers in the numeric photometric columns.

```python
# Check for missing values
print(df.isnull().sum())

# Drop or impute missing rows (illustrative — choose the approach that fits your data)
df = df.dropna()

# Example IQR-based outlier flagging on a numeric feature column
numeric_cols = ["u", "g", "r", "i", "z", "redshift"]

def flag_outliers_iqr(series):
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return (series < lower) | (series > upper)

outlier_mask = pd.DataFrame({c: flag_outliers_iqr(df[c]) for c in numeric_cols if c in df.columns})
print("Outlier counts per column:")
print(outlier_mask.sum())
```

## 3. Correlation analysis

Look at how the numeric features relate to one another before modeling.

```python
corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation matrix of photometric features (illustrative)")
plt.show()
```

## 4. Distributions and feature-target relationships

Compare feature distributions across the target classes (e.g. GALAXY / STAR / QSO).

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, ["redshift", "u", "g"]):
    sns.boxplot(data=df, x="class", y=col, ax=ax)
    ax.set_title(f"{col} by class (illustrative)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 4))
sns.histplot(data=df, x="redshift", hue="class", bins=40, kde=True)
plt.title("Redshift distribution by class (illustrative)")
plt.show()
```
